```python
house = {
    "Room A": "Dirty",
    "Room B": "Dirty"
}

location = "Room A"
battery = 10
energy_used = 0

while True:

    if house["Room A"] == "Clean" and house["Room B"] == "Clean":
        print("All rooms are clean!")
        print("Returning to charging dock...")
        break

    if house[location] == "Dirty":
        print("Dirt detected in", location)
        print("Sucking dirt...")

        house[location] = "Clean"
        battery -= 2
        energy_used += 2

    if location == "Room A":
        print("Moving Right")
        location = "Room B"

    elif location == "Room B":
        print("Moving Left")
        location = "Room A"

    battery -= 1
    energy_used += 1

    print("Location:", location)
    print("Battery:", battery)

    if battery <= 0:
        print("Battery low - Returning to charger")
        break

print("Energy Used:", energy_used)
print("Final House Status:", house)
```
